```python
from datetime import date
historique = []
```

```python
def montant_positif():
    while True:
        amount = input("Please enter a positive amount")
        try:
            number = float(amount)

            if number >= 0:
                print(f"Perfect, you have chosen: {number} as amount")
                return number
            else:
                print("Error: The amount must be positive.")
        except:
            print("Error, please enter a number not a text")
            
        return number
```

```python
def type_revenu():
    print("Quel est le type de revenu ?")
    print("1 - Salaire")
    print("2 - Freelance")

    choix = input("Tapez le numéro de votre choix (1 ou 2) : ")

    # On définit la description en fonction du numéro choisi
    if choix == "1":
        description = "Salaire"
    elif choix == "2":
        description = "Freelance"
    else:
        description = "Autre" 

    # On affiche le résultat
    return print("Description enregistrée :", description)
```

```python
def get_date():
    current_month = date.today().month
    return current_month
```

```python
def ajout_revenu():
    amount = montant_positif()
    description = type_revenu()
    date_month = get_date()

    revenu_entry = {
            "montant": amount,
            "description": description,
            "date": date_month
        }
    historique.append(revenu_entry)

ajout_revenu()
print(historique)
```
